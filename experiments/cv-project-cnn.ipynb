```python
# Import all required libraries
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import cv2
import os
import numpy as np
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
from tqdm import tqdm

# Check for GPU availability
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Dataset class
class SRDataset(Dataset):
    def __init__(self, data_tuples):
        self.data = data_tuples
        
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        lr_img, hr_img = self.data[idx]
        
        # Convert BGR to RGB
        lr_img = cv2.cvtColor(lr_img, cv2.COLOR_BGR2RGB)
        hr_img = cv2.cvtColor(hr_img, cv2.COLOR_BGR2RGB)
        
        # Normalize to [0, 1] and convert to float32
        lr_img = lr_img.astype('float32') / 255.0
        hr_img = hr_img.astype('float32') / 255.0
        
        # Convert to PyTorch tensors and permute to [C, H, W]
        lr_img = torch.from_numpy(lr_img).permute(2, 0, 1)
        hr_img = torch.from_numpy(hr_img).permute(2, 0, 1)
        
        return lr_img, hr_img

# Data loading function
def load_data(lr_path, hr_path):
    data_tuples = []
    for filename in os.listdir(lr_path):
        lr_img_path = os.path.join(lr_path, filename)
        hr_img_path = os.path.join(hr_path, filename)
        
        lr_img = cv2.imread(lr_img_path)
        hr_img = cv2.imread(hr_img_path)
        
        if lr_img is not None and hr_img is not None:
            data_tuples.append((lr_img, hr_img))
    
    return data_tuples

# CNN Model
class SuperResolutionCNN(nn.Module):
    def __init__(self):
        super(SuperResolutionCNN, self).__init__()
        
        # Feature extraction
        self.features = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=9, padding=4),
            nn.ReLU(),
            nn.Conv2d(64, 32, kernel_size=1, padding=0),
            nn.ReLU()
        )
        
        # Upsampling blocks
        self.upsample = nn.Sequential(
            nn.ConvTranspose2d(32, 32, kernel_size=3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(32, 32, kernel_size=3, stride=2, padding=1, output_padding=1),
            nn.ReLU()
        )
        
        # Reconstruction
        self.reconstruction = nn.Conv2d(32, 3, kernel_size=5, padding=2)
        
    def forward(self, x):
        x = self.features(x)
        x = self.upsample(x)
        x = self.reconstruction(x)
        return torch.sigmoid(x)

# Training function
def train_model(model, train_loader, test_loader, epochs=50, lr=0.001):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    
    train_losses = []
    test_losses = []
    best_loss = float('inf')
    
    for epoch in range(epochs):
        model.train()
        running_loss = 0.0
        
        for lr_imgs, hr_imgs in tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}"):
            lr_imgs = lr_imgs.to(device)
            hr_imgs = hr_imgs.to(device)
            
            optimizer.zero_grad()
            outputs = model(lr_imgs)
            loss = criterion(outputs, hr_imgs)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item()
        
        avg_train_loss = running_loss / len(train_loader)
        train_losses.append(avg_train_loss)
        
        # Validation
        model.eval()
        test_loss = 0.0
        with torch.no_grad():
            for lr_imgs, hr_imgs in test_loader:
                lr_imgs = lr_imgs.to(device)
                hr_imgs = hr_imgs.to(device)
                outputs = model(lr_imgs)
                test_loss += criterion(outputs, hr_imgs).item()
        
        avg_test_loss = test_loss / len(test_loader)
        test_losses.append(avg_test_loss)
        
        print(f"Epoch {epoch+1}/{epochs} - Train Loss: {avg_train_loss:.4f}, Test Loss: {avg_test_loss:.4f}")
        
        # Save best model
        if avg_test_loss < best_loss:
            best_loss = avg_test_loss
            torch.save(model.state_dict(), 'best_srcnn_model.pth')
    
    return train_losses, test_losses

# Main execution
if __name__ == "__main__":
    # Paths to your dataset
    lr_path = "/kaggle/input/cv-project-data/low_resolution"
    hr_path = "/kaggle/input/cv-project-data/high_resolution"
    
    # Load and split data
    print("Loading data...")
    data_tuples = load_data(lr_path, hr_path)
    train_tuples, test_tuples = train_test_split(data_tuples, test_size=0.1, random_state=42)
    
    # Create datasets
    print("Creating datasets...")
    train_dataset = SRDataset(train_tuples)
    test_dataset = SRDataset(test_tuples)
    
    # Create dataloaders
    train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
    test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)
    
    # Initialize model
    print("Initializing model...")
    model = SuperResolutionCNN().to(device)
    
    # Train the model
    print("Starting training...")
    train_losses, test_losses = train_model(model, train_loader, test_loader, epochs=100)
    
    # Plot results
    plt.figure(figsize=(10, 5))
    plt.plot(train_losses, label='Training Loss')
    plt.plot(test_losses, label='Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    plt.title('Training and Validation Loss')
    plt.show()
```

```python
# Visualization function
def visualize_results(model, test_loader, num_samples=3):
    model.eval()
    with torch.no_grad():
        # Get a batch of test data
        for lr_imgs, hr_imgs in test_loader:
            lr_imgs = lr_imgs.to(device)
            hr_imgs = hr_imgs.to(device)
            
            # Generate predictions
            pred_imgs = model(lr_imgs)
            
            # Convert tensors to numpy arrays
            lr_imgs = lr_imgs.cpu().numpy()
            hr_imgs = hr_imgs.cpu().numpy()
            pred_imgs = pred_imgs.cpu().numpy()
            
            # Plot results for the first few samples
            for i in range(min(num_samples, len(lr_imgs))):
                plt.figure(figsize=(15, 5))
                
                # Low resolution image
                plt.subplot(1, 3, 1)
                plt.imshow(np.clip(lr_imgs[i].transpose(1, 2, 0), 0, 1))
                plt.title('Low Resolution')
                plt.axis('off')
                
                # Model prediction
                plt.subplot(1, 3, 2)
                plt.imshow(np.clip(pred_imgs[i].transpose(1, 2, 0), 0, 1))
                plt.title('Model Prediction')
                plt.axis('off')
                
                # High resolution (ground truth)
                plt.subplot(1, 3, 3)
                plt.imshow(np.clip(hr_imgs[i].transpose(1, 2, 0), 0, 1))
                plt.title('High Resolution (Ground Truth)')
                plt.axis('off')
                
                plt.show()
                
                # Calculate metrics
                pred_img = (np.clip(pred_imgs[i].transpose(1, 2, 0), 0, 1) * 255).astype('uint8')
                hr_img = (np.clip(hr_imgs[i].transpose(1, 2, 0), 0, 1) * 255).astype('uint8')
                
                current_psnr = psnr(hr_img, pred_img)
                current_ssim = ssim(hr_img, pred_img, multichannel=True, channel_axis=2)
                
                print(f"Sample {i+1} - PSNR: {current_psnr:.2f}, SSIM: {current_ssim:.4f}")
                print("="*50)
            
            break  # Only show one batch

# Load the best model for visualization
best_model = SuperResolutionCNN().to(device)
best_model.load_state_dict(torch.load('best_srcnn_model.pth'))

# Visualize results
print("Visualizing results...")
visualize_results(best_model, test_loader, num_samples=3)
```
